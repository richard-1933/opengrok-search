### Search Conditions

````python
import os

param_spsf = {
    "full": "/.*(sp|sf)\_[a-zA-Z]{3}\_.*/ -type:sh -type:jar -type:javaclass +path:/[a-zA-Z]{3}\_MP/",
    "path": " -\". prc\" -\". fnc\" -\". trg\" -\". tri\" -\". viw\" -\". tab\" -\". pck\" -\". tps\" -/develop -/branch -/IAL -/CNWEB"
}

param_rowid_java = {
    "full": "'row_id' OR 'row_id' type:java"
    # "path": ""
}

param_row_id_other = {
    "full": "'row_id' OR 'rowid' -type:java -type:jar -type:javaclass"
    # "path": ""
}
````

### 1. 透過java呼叫的sp/sf

``` json
    params = {
        "full": "/.*(sp|sf)\_[a-zA-Z]{3}\_.*/ -type:sh -type:jar -type:javaclass +path:/[a-zA-Z]{3}\_MP/",
        "path": " -\". prc\" -\". fnc\" -\". trg\" -\". tri\" -\". viw\" -\". tab\" -\". pck\" -\". tps\" -/develop -/branch -/IAL -/CNWEB"
    }

```

````python
import opengrok_util
import re, json

joined_records = set()
records = opengrok_util.code_scan(param_spsf, fetch_all=True)
matches = re.findall(r'<b>(.*?)</b>', json.dumps(records, indent=4, ensure_ascii=False), flags=re.I)
# matches = re.findall(r'<b>(.[sp|sf]_[a-zA-Z]{3}_.*)</b>', json.dumps(records, indent=4, ensure_ascii=False), flags=re.I)
unique_matches = set(matches)
for item in unique_matches:
    joined_records.add(re.findall(r".[sp|sf]_[a-zA-Z]{3}_[a-zA-Z]*[0-9]*", item, re.I)[0])

print(f"Total Count: {len(joined_records)}")
# joined_records

````

### 3. 內含row_id / rowid關鍵字的程式 (db objs、shell.....)

````python
import opengrok_util
import re, json

records = opengrok_util.code_scan(param_row_id_other, fetch_all=False)
````

### Save to Excel

````python
import pandas as pd
import os
data = []
for file_path, entries in records.items():
    for entry in entries:
        dataItem = {"file_path": file_path, **entry}
        data.append(dataItem)
df = pd.DataFrame(data)
````
