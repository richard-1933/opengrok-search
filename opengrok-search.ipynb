```python
# importation here
import pprint, requests, re,json
from requests.packages.urllib3.exceptions import InsecureRequestWarning

requests.packages.urllib3.disable_warnings(InsecureRequestWarning)
```

```python
# define variables
records_per_call = 100
url = "https://tpedk02t.wanhai.com/opengrok/api/v1/search"

headers = {
    "Accept": "application/json"
}
```

```python
def fetch_scan_result(start_record_idx :int, fetch_records :int):
    params = {
        "full": "/.*(sp|sf)\_[a-zA-Z]{3}\_.*/ -type:sh -type:jar -type:javaclass +path:/[a-zA-Z]{3}\_MP/",
        "path": " -\". prc\" -\". fnc\" -\". trg\" -\". tri\" -\". viw\" -\". tab\" -\". pck\" -\". tps\" -/develop -/branch -/IAL -/CNWEB",
        "start": start_record_idx,
        "maxresults": fetch_records
    }
    response = requests.get(url, params=params , headers=headers, verify=False)
    json_obj = response.json()
    return json_obj['results']
```

```python
keep_fetching = True
joined_records = set()
start_record = 0
while keep_fetching:
    records = fetch_scan_result(start_record, records_per_call)
    if len(records) == 0:
        keep_fetching = False
        break
        
    matches = re.findall(r'<b>(.*?)</b>', json.dumps(records, indent=4, ensure_ascii=False))
    unique_matches = set(matches)
    joined_records = joined_records | unique_matches
    start_record += records_per_call
pprint.pprint(joined_records)
print(f"TTL Count: {len(joined_records)}")
```
